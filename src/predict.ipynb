# Predict AQI

Load the model produced by `train.ipynb`, then predict a batch CSV with matching pollutant and city columns.


```python
from pathlib import Path
import joblib
import pandas as pd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "models/random_forest_aqi.pkl").exists())
artifact = joblib.load(ROOT / "models/random_forest_aqi.pkl")
model, feature_columns = artifact["model"], artifact["features"]
```

```python
# Set this to the input CSV path. It must contain the trained predictor columns.
input_path = ROOT / "data/processed/input.csv"
output_path = ROOT / "data/processed/predictions.csv"
if input_path.exists():
    incoming = pd.read_csv(input_path)
    incoming["Predicted_AQI"] = model.predict(incoming.reindex(columns=feature_columns))
    incoming.to_csv(output_path, index=False)
    print(f"Saved predictions to {output_path}")
    incoming.head()
else:
    print(f"Add an input CSV with columns {feature_columns} at {input_path}")
```
