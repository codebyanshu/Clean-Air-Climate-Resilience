# AQI prediction demo

Notebook-based demo for a single AQI estimate. Train the model with `src/train.ipynb` first.


```python
from pathlib import Path
import joblib
import pandas as pd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "models/random_forest_aqi.pkl").exists())
artifact = joblib.load(ROOT / "models/random_forest_aqi.pkl")
model, features = artifact["model"], artifact["features"]
```

```python
sample = {
    "City": "Delhi", "PM2.5": 150, "PM10": 250, "NO": 20, "NO2": 60,
    "NOx": 80, "NH3": 20, "CO": 1.5, "SO2": 20, "O3": 80,
    "Benzene": 2, "Toluene": 5,
}
row = pd.DataFrame([sample]).reindex(columns=features)
predicted_aqi = float(model.predict(row)[0])
print(f"Predicted AQI: {predicted_aqi:.1f}")
```
